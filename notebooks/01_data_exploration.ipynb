```python
#preparing cleaning process
df_clean = df.copy()

# Check the dimensions of both datasets
print("Original dataset:", df.shape)
print("Working dataset:", df_clean.shape)
```

```python
#checking the working copy
print("Are both datasets identical?",
      df.equals(df_clean))
```

```python
#finding stockcode belonging to rows with misisng description
missing_description_codes = (
    df_clean.loc[df_clean["Description"].isna(), "StockCode"]
    .unique()
)

print("Unique StockCodes with missing descriptions:",
      len(missing_description_codes))
```

Checking for recoverable descriptions

```python
# Find rows where descriptions are available
available_descriptions = df_clean[
    df_clean["Description"].notna()
]

# Find descriptions associated with the missing-description StockCodes
recoverable_descriptions = available_descriptions[
    available_descriptions["StockCode"].isin(
        missing_description_codes
    )
]

print("Records with available descriptions for these codes:",
      recoverable_descriptions.shape[0])

print("Unique StockCodes with available descriptions:",
      recoverable_descriptions["StockCode"].nunique())
```

```python
# Count unique descriptions for each StockCode
description_counts = (
    available_descriptions
    .groupby("StockCode")["Description"]
    .nunique()
)

# Identify StockCodes with multiple descriptions
multiple_descriptions = description_counts[
    description_counts > 1
]

print("StockCodes with multiple descriptions:",
      multiple_descriptions.shape[0])

display(multiple_descriptions.head(20))
```

Separating safe and ambiguous Stockcodes


```python
# Codes with exactly one available description
consistent_codes = description_counts[
    description_counts == 1
]

# Codes with multiple available descriptions
ambiguous_codes = description_counts[
    description_counts > 1
]

print("Codes with one consistent description:",
      len(consistent_codes))

print("Codes with multiple descriptions:",
      len(ambiguous_codes))
```

identifying codes that are safe candidiates for filling

```python
# Find missing-description StockCodes with exactly one description elsewhere
safe_codes = consistent_codes[
    consistent_codes.index.isin(missing_description_codes)
]

# Find missing-description StockCodes with multiple descriptions elsewhere
ambiguous_missing_codes = ambiguous_codes[
    ambiguous_codes.index.isin(missing_description_codes)
]

print("Missing-description codes with one consistent description:",
      len(safe_codes))

print("Missing-description codes with multiple descriptions:",
      len(ambiguous_missing_codes))

print("Missing-description codes with no available description:",
      960 - len(safe_codes) - len(ambiguous_missing_codes))
```

Idntifying safe candidates

```python
# Missing description codes with exactly one known description
safe_codes=consistent_codes[
    consistent_codes.index.isin(missing_description_codes)
]

#missing description codes with multiple known descriptions
ambiguous_missing_codes=ambiguous_codes[
    ambiguous_codes.index.isin(missing_description_codes)
]

#codes with no known description
uncoverable_count=(
    len(missing_description_codes)
    -len(safe_codes)
    -len(ambiguous_missing_codes)
)
print("Safe Candidiates:",len(safe_codes))
print("Ambiguoys codes:",len(ambiguous_missing_codes))
print("Codes with no available description:",uncoverable_count)
```

Previewing the descriptions we can recover


```python
# Create a mapping of safe StockCodes to their descriptions
safe_description_map = (
    available_descriptions[
        available_descriptions["StockCode"].isin(safe_codes.index)
    ]
    .drop_duplicates(subset="StockCode")
    .set_index("StockCode")["Description"]
)

# Preview missing descriptions that can potentially be recovered
safe_missing_rows = df_clean[
    df_clean["Description"].isna()
    & df_clean["StockCode"].isin(safe_codes.index)
].copy()

safe_missing_rows["RecoveredDescription"] = (
    safe_missing_rows["StockCode"].map(safe_description_map)
)

print("Missing records eligible for recovery:",
      safe_missing_rows.shape[0])

display(
    safe_missing_rows[
        ["StockCode", "Description", "RecoveredDescription"]
    ].head(20)
)
```
